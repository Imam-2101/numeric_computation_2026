```python
# =============================================================
# Penyelesaian SPL dengan Metode Eliminasi Gauss
# (dengan partial pivoting) - siap dijalankan di Google Colab
# =============================================================
import numpy as np

EPSILON = 1e-12


def gauss_eliminasi(A, b, tampil_langkah=True):
    """Menyelesaikan A.x = b dengan eliminasi Gauss.
    Mengembalikan vektor x, atau None jika tidak ada solusi tunggal."""
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    # Bentuk matriks augmented [A | b]
    M = np.hstack((A, b.reshape(-1, 1)))
    np.set_printoptions(precision=4, suppress=True, linewidth=120)

    if tampil_langkah:
        print("Matriks augmented awal:")
        print(M, "\n")

    # ---------- Tahap 1: Eliminasi maju ----------
    for k in range(n - 1):
        # Partial pivoting: cari baris dengan |nilai| terbesar di kolom k
        p = k + np.argmax(np.abs(M[k:, k]))

        if abs(M[p, k]) < EPSILON:
            print("SPL tidak memiliki solusi tunggal (pivot = 0).")
            return None

        if p != k:
            M[[k, p]] = M[[p, k]]  # tukar baris
            if tampil_langkah:
                print(f"Tukar baris R{k+1} <-> R{p+1}")

        for i in range(k + 1, n):
            m = M[i, k] / M[k, k]
            M[i, k:] = M[i, k:] - m * M[k, k:]
            if tampil_langkah:
                print(f"R{i+1} <- R{i+1} - ({m:.4f}) * R{k+1}")

        if tampil_langkah:
            print(f"\nSetelah eliminasi kolom {k+1}:")
            print(M, "\n")

    if abs(M[n - 1, n - 1]) < EPSILON:
        print("SPL tidak memiliki solusi tunggal (pivot = 0).")
        return None

    # ---------- Tahap 2: Substitusi mundur ----------
    x = np.zeros(n)
    x[n - 1] = M[n - 1, n] / M[n - 1, n - 1]
    for i in range(n - 2, -1, -1):
        jumlah = np.dot(M[i, i + 1:n], x[i + 1:n])
        x[i] = (M[i, n] - jumlah) / M[i, i]

    return x


def baca_input():
    """Membaca SPL dari input pengguna."""
    n = int(input("Masukkan jumlah persamaan/variabel (n): "))
    print(f"\nMasukkan koefisien matriks A ({n} baris, pisahkan dengan spasi):")
    A = []
    for i in range(n):
        baris = list(map(float, input(f"  Baris {i+1}: ").split()))
        if len(baris) != n:
            raise ValueError(f"Baris harus berisi {n} angka.")
        A.append(baris)
    print(f"\nMasukkan vektor konstanta b ({n} angka, pisahkan dengan spasi):")
    b = list(map(float, input("  b: ").split()))
    if len(b) != n:
        raise ValueError(f"Vektor b harus berisi {n} angka.")
    return A, b


def main():
    print("=== SPL - Metode Eliminasi Gauss ===")
    print("1. Gunakan contoh soal")
    print("2. Input sendiri")
    pilihan = input("Pilih (1/2): ").strip()

    if pilihan == "2":
        A, b = baca_input()
    else:
        # Contoh:  2x + y - z  = 8
        #         -3x - y + 2z = -11
        #         -2x + y + 2z = -3      -> solusi: x=2, y=3, z=-1
        A = [[2, 1, -1],
             [-3, -1, 2],
             [-2, 1, 2]]
        b = [8, -11, -3]
        print("\nContoh soal yang digunakan:")
        for i in range(len(b)):
            print("  ", A[i], "| ", b[i])
        print()

    x = gauss_eliminasi(A, b)

    if x is not None:
        print("=== HASIL ===")
        for i, nilai in enumerate(x, start=1):
            print(f"x{i} = {nilai:.6f}")

        # Verifikasi dengan NumPy
        print("\nVerifikasi (numpy.linalg.solve):", np.linalg.solve(A, b))
        print("Residual ||A.x - b|| =", np.linalg.norm(np.array(A) @ x - np.array(b)))


main()
```

# Penyelesaian Sistem Persamaan Linear (SPL)
## Metode Eliminasi Gauss dengan Partial Pivoting

Program ini digunakan untuk menyelesaikan Sistem Persamaan Linear (SPL) menggunakan metode Eliminasi Gauss. Proses penyelesaian dibagi menjadi dua langkah utama, yaitu eliminasi maju dan substitusi mundur.

Program menggunakan library NumPy untuk membantu operasi matriks dan melakukan verifikasi hasil.

```python
import numpy as np

EPSILON = 1e-12
np.set_printoptions(precision=4, suppress=True, linewidth=120)
```

Kode ini mengimpor library NumPy untuk melakukan perhitungan numerik dan operasi matriks. Variabel EPSILON digunakan sebagai batas toleransi untuk mendeteksi nilai pivot yang mendekati nol.

```python
def baca_input():
    n = int(input("Masukkan jumlah persamaan/variabel (n): "))

    print(f"\nMasukkan koefisien matriks A ({n} baris):")
    A = []

    for i in range(n):
        baris = list(map(float, input(f"Baris {i+1}: ").split()))

        if len(baris) != n:
            raise ValueError(f"Baris harus berisi {n} angka.")

        A.append(baris)

    print(f"\nMasukkan vektor konstanta b ({n} angka):")
    b = list(map(float, input("b: ").split()))

    if len(b) != n:
        raise ValueError(f"Vektor b harus berisi {n} angka.")

    return A, b
```

Fungsi baca_input() digunakan untuk menerima jumlah persamaan, koefisien matriks A, dan konstanta b dari pengguna. Program memeriksa jumlah angka yang dimasukkan agar sesuai dengan ukuran SPL.

## Langkah 1: Eliminasi Maju

Eliminasi maju bertujuan mengubah matriks augmented [A | b] menjadi matriks segitiga atas. Proses dilakukan dengan menghilangkan koefisien di bawah elemen pivot menjadi nol.

Pada program ini digunakan teknik partial pivoting, yaitu memilih baris dengan nilai absolut terbesar pada kolom pivot, kemudian menukar baris jika diperlukan. Teknik ini membantu mengurangi kesalahan numerik selama perhitungan.

```python
def eliminasi_maju(A, b, tampil_langkah=True):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    n = len(b)
    M = np.hstack((A, b.reshape(-1, 1)))

    if tampil_langkah:
        print("Matriks augmented awal:")
        print(M, "\n")

    for k in range(n - 1):
        # Partial pivoting
        p = k + np.argmax(np.abs(M[k:, k]))

        if abs(M[p, k]) < EPSILON:
            print("SPL tidak memiliki solusi tunggal.")
            return None

        if p != k:
            M[[k, p]] = M[[p, k]]

            if tampil_langkah:
                print(f"Tukar baris R{k+1} <-> R{p+1}")

        # Eliminasi elemen di bawah pivot
        for i in range(k + 1, n):
            m = M[i, k] / M[k, k]
            M[i, k:] = M[i, k:] - m * M[k, k:]

            if tampil_langkah:
                print(f"R{i+1} <- R{i+1} - ({m:.4f}) * R{k+1}")

        if tampil_langkah:
            print(f"\nSetelah eliminasi kolom {k+1}:")
            print(M, "\n")

    # Pemeriksaan pivot terakhir
    if abs(M[n - 1, n - 1]) < EPSILON:
        print("SPL tidak memiliki solusi tunggal.")
        return None

    if tampil_langkah:
        print("Matriks segitiga atas:")
        print(M, "\n")

    return M
```

Fungsi eliminasi_maju() membentuk matriks augmented, menentukan pivot dengan partial pivoting, menukar baris apabila diperlukan, dan menghilangkan elemen di bawah pivot. Hasil akhirnya berupa matriks segitiga atas yang akan digunakan pada langkah substitusi mundur.

## Langkah 2: Substitusi Mundur

Substitusi mundur bertujuan memperoleh nilai setiap variabel dari matriks segitiga atas yang dihasilkan pada langkah eliminasi maju.

Perhitungan dimulai dari variabel terakhir, kemudian dilanjutkan ke variabel sebelumnya hingga semua nilai variabel ditemukan.

```python
def substitusi_mundur(M):
    n = M.shape[0]
    x = np.zeros(n)

    # Menghitung variabel terakhir
    x[n - 1] = M[n - 1, n] / M[n - 1, n - 1]

    # Menghitung variabel sebelumnya
    for i in range(n - 2, -1, -1):
        jumlah = np.dot(M[i, i + 1:n], x[i + 1:n])
        x[i] = (M[i, n] - jumlah) / M[i, i]

    return x
```

Fungsi substitusi_mundur() menghitung nilai variabel dari baris terakhir menuju baris pertama. Nilai variabel yang sudah diperoleh digunakan untuk menghitung variabel sebelumnya hingga seluruh solusi SPL ditemukan.

## Menjalankan Program

Pada bagian ini, pengguna dapat memilih contoh soal yang tersedia atau memasukkan SPL sendiri. Matriks kemudian diproses menggunakan eliminasi maju dan substitusi mundur untuk mendapatkan hasil akhir.

```python
def main():
    print("=== SPL - Metode Eliminasi Gauss ===")
    print("1. Gunakan contoh soal")
    print("2. Input sendiri")

    pilihan = input("Pilih (1/2): ").strip()

    if pilihan == "2":
        A, b = baca_input()
    else:
        A = [
            [2, 1, -1],
            [-3, -1, 2],
            [-2, 1, 2]
        ]

        b = [8, -11, -3]

        print("\nContoh soal:")
        for i in range(len(b)):
            print(A[i], "|", b[i])

    # Langkah 1: Eliminasi maju
    M = eliminasi_maju(A, b)

    if M is None:
        return

    # Langkah 2: Substitusi mundur
    x = substitusi_mundur(M)

    print("=== HASIL ===")
    for i, nilai in enumerate(x, start=1):
        print(f"x{i} = {nilai:.6f}")

    # Verifikasi hasil
    print("\nVerifikasi (numpy.linalg.solve):",
          np.linalg.solve(A, b))

    print("Residual ||A.x - b|| =",
          np.linalg.norm(np.array(A) @ x - np.array(b)))


main()
```

Fungsi main() mengatur jalannya program, mulai dari pemilihan input, pemanggilan eliminasi maju, hingga substitusi mundur. Program menampilkan nilai setiap variabel dan melakukan verifikasi menggunakan numpy.linalg.solve(). Nilai residual digunakan untuk memeriksa apakah hasil perhitungan memenuhi sistem persamaan yang diberikan.
