```python
# =============================================================
# Penyelesaian SPL dengan Metode Eliminasi Gauss
# (dengan partial pivoting) - siap dijalankan di Google Colab
# =============================================================
import numpy as np

EPSILON = 1e-12


def gauss_eliminasi(A, b, tampil_langkah=True):
    """Menyelesaikan A.x = b dengan eliminasi Gauss.
    Mengembalikan vektor x, atau None jika tidak ada solusi tunggal."""
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    # Bentuk matriks augmented [A | b]
    M = np.hstack((A, b.reshape(-1, 1)))
    np.set_printoptions(precision=4, suppress=True, linewidth=120)

    if tampil_langkah:
        print("Matriks augmented awal:")
        print(M, "\n")

    # ---------- Tahap 1: Eliminasi maju ----------
    for k in range(n - 1):
        # Partial pivoting: cari baris dengan |nilai| terbesar di kolom k
        p = k + np.argmax(np.abs(M[k:, k]))

        if abs(M[p, k]) < EPSILON:
            print("SPL tidak memiliki solusi tunggal (pivot = 0).")
            return None

        if p != k:
            M[[k, p]] = M[[p, k]]  # tukar baris
            if tampil_langkah:
                print(f"Tukar baris R{k+1} <-> R{p+1}")

        for i in range(k + 1, n):
            m = M[i, k] / M[k, k]
            M[i, k:] = M[i, k:] - m * M[k, k:]
            if tampil_langkah:
                print(f"R{i+1} <- R{i+1} - ({m:.4f}) * R{k+1}")

        if tampil_langkah:
            print(f"\nSetelah eliminasi kolom {k+1}:")
            print(M, "\n")

    if abs(M[n - 1, n - 1]) < EPSILON:
        print("SPL tidak memiliki solusi tunggal (pivot = 0).")
        return None

    # ---------- Tahap 2: Substitusi mundur ----------
    x = np.zeros(n)
    x[n - 1] = M[n - 1, n] / M[n - 1, n - 1]
    for i in range(n - 2, -1, -1):
        jumlah = np.dot(M[i, i + 1:n], x[i + 1:n])
        x[i] = (M[i, n] - jumlah) / M[i, i]

    return x


def baca_input():
    """Membaca SPL dari input pengguna."""
    n = int(input("Masukkan jumlah persamaan/variabel (n): "))
    print(f"\nMasukkan koefisien matriks A ({n} baris, pisahkan dengan spasi):")
    A = []
    for i in range(n):
        baris = list(map(float, input(f"  Baris {i+1}: ").split()))
        if len(baris) != n:
            raise ValueError(f"Baris harus berisi {n} angka.")
        A.append(baris)
    print(f"\nMasukkan vektor konstanta b ({n} angka, pisahkan dengan spasi):")
    b = list(map(float, input("  b: ").split()))
    if len(b) != n:
        raise ValueError(f"Vektor b harus berisi {n} angka.")
    return A, b


def main():
    print("=== SPL - Metode Eliminasi Gauss ===")
    print("1. Gunakan contoh soal")
    print("2. Input sendiri")
    pilihan = input("Pilih (1/2): ").strip()

    if pilihan == "2":
        A, b = baca_input()
    else:
        # Contoh:  2x + y - z  = 8
        #         -3x - y + 2z = -11
        #         -2x + y + 2z = -3      -> solusi: x=2, y=3, z=-1
        A = [[2, 1, -1],
             [-3, -1, 2],
             [-2, 1, 2]]
        b = [8, -11, -3]
        print("\nContoh soal yang digunakan:")
        for i in range(len(b)):
            print("  ", A[i], "| ", b[i])
        print()

    x = gauss_eliminasi(A, b)

    if x is not None:
        print("=== HASIL ===")
        for i, nilai in enumerate(x, start=1):
            print(f"x{i} = {nilai:.6f}")

        # Verifikasi dengan NumPy
        print("\nVerifikasi (numpy.linalg.solve):", np.linalg.solve(A, b))
        print("Residual ||A.x - b|| =", np.linalg.norm(np.array(A) @ x - np.array(b)))


main()
```

## Langkah 2: Membuat "Mesin Penebak" ⚙️

Di sini kita membuat **mesin** bernama `gauss_eliminasi`.
Mesin ini belum jalan. Kita hanya **merakitnya** dulu.

Mesin ini bekerja dalam **2 tahap**:

### Tahap 1: Beres-beres (Eliminasi maju) 🧹
Kita susun teka-teki itu dalam sebuah tabel. Lalu kita bereskan
tabel itu supaya bentuknya seperti **tangga**: angka di kiri bawah dibuat jadi 0.

Sebelum:            Sesudah (bentuk tangga):

    x x x | x           x x x | x
    x x x | x           0 x x | x
    x x x | x           0 0 x | x

Kenapa harus jadi tangga? Karena baris paling bawah jadi **sangat mudah**:
hanya tersisa satu angka rahasia (z).

Cara membereskan tiap kolom:
1. **Pilih ketua** (disebut *pivot*): angka yang paling besar di kolom itu.
   Kita pilih yang besar karena komputer lebih teliti kalau memakai angka besar.
2. **Tukar tempat** baris itu supaya si ketua duduk paling atas.
3. **Hapus angka di bawah ketua** dengan cara mengurangi baris.

### Tahap 2: Tebak dari bawah (Substitusi mundur) 🔙
Karena sudah jadi tangga, kita mulai dari baris paling bawah:
- Baris bawah hanya punya z, jadi **z langsung ketemu**.
- Pakai z untuk mencari **y**.
- Pakai y dan z untuk mencari **x**.

Seperti menyusun menara balok: dari bawah dulu, baru ke atas.

```python
def gauss_eliminasi(A, b, tampil_langkah=True):
    """Menyelesaikan A.x = b dengan eliminasi Gauss + partial pivoting.
    Mengembalikan vektor x, atau None jika tidak ada solusi tunggal."""

    A = np.array(A, dtype=float)  # ubah A jadi array NumPy bertipe desimal
    b = np.array(b, dtype=float)  # ubah b jadi array NumPy bertipe desimal
    n = len(b)                    # jumlah persamaan/variabel (3)

    # Validasi: A harus persegi n x n agar sesuai dengan b
    if A.shape != (n, n):
        raise ValueError("Matriks A harus persegi dan sesuai ukuran b.")

    # Matriks augmented: gabungkan A (3x3) dan b (3x1) menjadi M (3x4)
    # b.reshape(-1, 1) mengubah b menjadi kolom agar bisa digabung
    M = np.hstack((A, b.reshape(-1, 1)))

    if tampil_langkah:
        print("Matriks augmented awal:")
        print(M, "\n")

    # ---------- Tahap 1: Eliminasi maju ----------
    for k in range(n - 1):  # k = indeks kolom pivot (0 dan 1)

        # Partial pivoting: cari baris dengan |nilai| terbesar mulai baris k ke bawah.
        # np.argmax memberi posisi relatif, jadi ditambah k untuk mendapat indeks asli.
        p = k + np.argmax(np.abs(M[k:, k]))

        # Jika pivot terbesar hampir nol, SPL tidak punya solusi tunggal
        if abs(M[p, k]) < EPSILON:
            print("SPL tidak memiliki solusi tunggal (pivot = 0).")
            return None

        # Jika pivot terbesar bukan di baris k, tukar baris k dengan baris p
        if p != k:
            M[[k, p]] = M[[p, k]]  # penukaran dua baris sekaligus
            if tampil_langkah:
                print(f"Tukar baris R{k+1} <-> R{p+1}")

        # Nolkan semua angka di bawah pivot pada kolom k
        for i in range(k + 1, n):
            m = M[i, k] / M[k, k]                 # faktor pengali
            M[i, k:] = M[i, k:] - m * M[k, k:]    # R_i <- R_i - m * R_k
            if tampil_langkah:
                print(f"R{i+1} <- R{i+1} - ({m:.4f}) * R{k+1}")

        if tampil_langkah:
            print(f"\nSetelah eliminasi kolom {k+1}:")
            print(M, "\n")

    # Cek pivot terakhir (baris 3, kolom 3). Jika nol, tidak ada solusi tunggal
    if abs(M[n - 1, n - 1]) < EPSILON:
        print("SPL tidak memiliki solusi tunggal (pivot = 0).")
        return None

    # ---------- Tahap 2: Substitusi mundur ----------
    x = np.zeros(n)  # wadah hasil [x, y, z], awalnya berisi nol

    # Variabel terakhir (z) langsung didapat dari baris terakhir
    x[n - 1] = M[n - 1, n] / M[n - 1, n - 1]

    # Variabel lainnya dihitung dari bawah ke atas (i = 1, lalu 0)
    for i in range(n - 2, -1, -1):
        jumlah = np.dot(M[i, i + 1:n], x[i + 1:n])  # jumlah koefisien * variabel yang sudah diketahui
        x[i] = (M[i, n] - jumlah) / M[i, i]         # (konstanta - jumlah) / koefisien diagonal

    return x  # kembalikan solusi [x, y, z]
```

## Langkah 5: Lihat Hasil dan Periksa Jawaban ✅

Mesin sudah menemukan x, y, z. Tapi apakah jawabannya benar?
Seperti anak sekolah yang mengerjakan PR, kita **periksa ulang** dengan 2 cara:

1. **Cara pembanding:** kita tanya kalkulator pintar NumPy
   (`numpy.linalg.solve`). Kalau jawabannya sama, berarti kita benar.
2. **Residual** (sisa kesalahan): kita masukkan jawaban kita ke teka-teki,
   lalu hitung seberapa meleset hasilnya dari yang seharusnya.
   Kalau meleset 0, berarti sempurna.

```python
if x is not None:  # tampilkan hanya jika solusi tunggal ditemukan
    nama = ["x", "y", "z"]  # nama variabel sesuai urutan hasil

    print("=== HASIL ===")
    for n_var, nilai in zip(nama, x):   # pasangkan nama variabel dengan nilainya
        print(f"{n_var} = {nilai:.6f}")

    # Bandingkan dengan fungsi bawaan NumPy sebagai pengecekan
    print("\nVerifikasi (numpy.linalg.solve):", np.linalg.solve(A, b))

    # Residual = panjang (norma) vektor selisih A.x - b, idealnya mendekati 0
    print("Residual ||A.x - b|| =", np.linalg.norm(np.array(A) @ x - np.array(b)))
```
